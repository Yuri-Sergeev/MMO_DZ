## Предобработка

1. 'consumption_mixed': смешанный расход топлива. Значение -1 некорректно. Данные с таким значением удалим.
2. 'max_speed': максимальная скорость. Максимальная скорость не может быть отрицательной. Данные с таким значением удалим.
4. 'weight': масса автомобиля. Не может быть отрицательной. Данные с таким значением удалим.
5. 'configurations_back_wheel_base': величина задней колесной базы. Не может быть отрицательной. Данные с таким значением удалим.

```python
df = df.loc[df['consumption-mixed'] > 0]
df = df.loc[df['max-speed'] > 0]
df = df.loc[df['weight'] > 0]
df = df.loc[df['configurations_back-wheel-base'] > 0]
```

```python
df.shape
```

```python
X = df.drop(["target"], axis=1)
y = df[["target"]]
```

```python
X = df.drop(["target",'index'], axis=1)
y = df[["target"]]
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.33, random_state=42)
lin = LinearRegression().fit(train_X, train_y)
```

```python
lin.score(train_X, train_y), lin.score(test_X, test_y)
```

```python
mean_squared_error(lin.predict(train_X), train_y), mean_squared_error(lin.predict(test_X), test_y)
```
