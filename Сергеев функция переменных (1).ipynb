```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set();

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
```

## Функция одной переменной

```python
def f(x):
    return 3*x**2
```

```python
def df(x):
    return 6*x
```

```python
x = np.arange(-10,11)
plt.plot(x, f(x))
```

```python
x0 = 5
l = 0.1
res=[]
x1 = x0 - l*df(x0)
```

```python
norm = ((x1 - x0)**2 + (f(x1) - f(x0))**2)**0.5
while norm > 0.00000001:
    small_list = [x0, f(x0)]
    res.append(small_list)  
    x0 = x1
    x1 = x0 - l*df(x0)
    norm = ((x1 - x0)**2 + (f(x1) - f(x0))**2)**0.5
```

```python
res
```

```python
a = list(map(lambda sublist: sublist[0], res))
b = list(map(lambda sublist: sublist[1], res))
```

```python

x = np.arange(-5.1,5.1,0.1)
fig, ax = plt.subplots() 
ax.plot(x, f(x), label='y1')
ax.plot(a, b, label='y2',marker='o')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Два графика на одном')
```

## Функция двух переменных

```python
def f(x):
    return 3*x[0]**2 + 5*x[1]**2
```

```python
def df(x):
    return np.array([6*x[0], 10*x[1]])
```

```python
x = [np.arange(-10,11), np.arange(-10,11)]
```

```python
X = np.meshgrid(x[0],x[1])
Z = f(X)
fig = plt.figure(figsize=(11, 10))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X[0], X[1], Z, cmap=cm.viridis)
plt.show()                    
```

```python
x0 = [5,10]
l = 0.1
res = []
```

```python
x0 = np.array([5,10], dtype=float)
l = 0.1
res = []
x1 = x0 - l*df(x0)
norm = np.linalg.norm(x1 - x0)
while norm > 0.00000001:
    small_list = [x0.copy(), f(x0)]
    res.append(small_list)
    x0 = x1.copy()
    x1 = x0 - l*df(x0)
    norm = np.linalg.norm(x1 - x0)
```

```python
res
```

```python
a = list(map(lambda sublist: sublist[0][0], res))
b = list(map(lambda sublist: sublist[0][1], res))
c = list(map(lambda sublist: sublist[1], res))
```

```python
X = np.meshgrid(x[0],x[1])
Z = f(X)
fig = plt.figure(figsize=(11, 10))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X[0], X[1], Z, cmap='viridis', alpha=0.3, linewidth=0)
ax.scatter(a, b, c, color='black', s=50)
ax.plot(a, b, c, color='black', marker='o', markersize=5, label='Траектория оптимизации')
ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_zlabel('f(x1, x2)')
ax.set_title('Градиентный спуск на 3D поверхности')
ax.legend()
plt.show()                 
```
